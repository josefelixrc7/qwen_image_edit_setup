```python
# Instalar y ejecutar cloudflared
!wget -q https://github.com/cloudflare/cloudflared/releases/latest/download/cloudflared-linux-amd64 -O cloudflared
!chmod +x cloudflared

import subprocess
import threading
import time

# Iniciar túnel en segundo plano
def start_tunnel():
    subprocess.Popen(['./cloudflared', 'tunnel', '--url', 'http://localhost:8188'])

tunnel_thread = threading.Thread(target=start_tunnel, daemon=True)
tunnel_thread.start()

print("⏳ Iniciando túnel... Busca la URL 'trycloudflare.com' en la salida arriba")
time.sleep(5)

# Iniciar ComfyUI
!python /mnt/comfyui/ComfyUI/main.py --listen --port 8188
```
