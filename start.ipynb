```python
# Instalar pinggy
%uv pip install pinggy -q

import pinggy
import threading
import time

# Iniciar túnel en segundo plano
def start_tunnel():
    tunnel = pinggy.start_tunnel(forwardto="localhost:8188")
    print(f"\n✅ Túnel activo: {tunnel.urls}")
    print(f"   🔗 Haz clic aquí para abrir ComfyUI: {tunnel.urls}\n")

tunnel_thread = threading.Thread(target=start_tunnel, daemon=True)
tunnel_thread.start()

time.sleep(3)

# Iniciar ComfyUI
!python /mnt/comfyui/main.py --listen --port 8188
```
